### Counting in Pandas

This notebook focuses only on **counting data in Pandas**.

You will learn:

- `len()`
- `count()`
- `value_counts()`
- `nunique()`
- Counting with conditions
- `groupby().count()`
- `groupby().size()`
- `pd.crosstab()`
- Titanic counting examples


### 1. Import Pandas and Create Sample Data

We will first use a small student dataset so the counting methods are easy to understand.


```python
import pandas as pd

# Create sample data
data = {
    "Name": ["pratiksha", "jina", "Chandrika", "Dipesha", "Elya", "Farana"],
    "Department": ["CSIT", "BCA", "CSIT", "BIM", "BCA", "CSIT"],
    "Age": [22, 21, 23, 24, 22, 25],
    "Score": [85, 72, 91, None, 68, 77],
    "City": ["Kathmandu", "Pokhara", "Kathmandu", "Butwal", "Pokhara", None]
}

# Convert dictionary to DataFrame
df = pd.DataFrame(data)

df
```

### 2. Count Total Rows with `len()`

`len(df)` counts the total number of rows.

It counts a row even if some values in that row are missing.


```python
# Count total number of rows
len(df)
```

```python
# QUESTION:
# Count the total number of students in the DataFrame.

# Write your answer below
#solution
# len(df) counts all rows
len(df)


```

### 3. Count Non-Missing Values with `count()`

`count()` counts only values that are not missing.

If a column has `NaN`, those missing values are not counted.


```python
# Count non-missing Score values
df["Score"].count()
```

```python
# QUESTION:
# Count the non-missing values in the City column.

# Write your answer below
# SOLUTION:
df["City"].count()

```

### 4. Count Non-Missing Values in Every Column

`df.count()` returns the number of non-missing values for each column.


```python
# Count non-missing values in every column
df.count()
```

### 5. Count Values Row by Row

Use `axis=1` to count non-missing values across each row.


```python
# Count non-missing values in every row
df.count(axis=1)
```

### 6. Count Frequency with `value_counts()`

`value_counts()` tells us how many times each unique value appears.


```python
# Count students in each Department
df["Department"].value_counts()
```

```python
# QUESTION:
# Count how many students come from each City.

# Write your answer below
# SOLUTION:
df["City"].value_counts()

```

### 7. Include Missing Values in `value_counts()`

By default, `value_counts()` ignores missing values.

Use `dropna=False` to include them.


```python
# Count City values including missing values
df["City"].value_counts(dropna=False)
```

```python
# QUESTION:
# Count Department values and include missing values if any.

# Write your answer below
# SOLUTION:
df["Department"].value_counts(dropna=False)


```

### 8. Count Unique Values with `nunique()`

`nunique()` tells us how many different values exist.


```python
# Count unique Departments
df["Department"].nunique()
```

```python
# QUESTION:
# Count how many unique City values are present.

# Write your answer below
# SOLUTION:
df["City"].nunique()
```

### 9. Count Rows with a Condition

A condition returns `True` or `False`.

In Pandas:

- `True` behaves like `1`
- `False` behaves like `0`

So `.sum()` can count how many rows satisfy the condition.


```python
# Count students whose Score is 80 or more
(df["Score"] >= 80).sum()
```

```python
# QUESTION:
# Count how many students are from the CSIT Department.

# Write your answer below
#solution
df[df["Department"] == "IT"].shape[0]

```

### 10. Count with Multiple Conditions

Use `&` when both conditions must be true.


```python
# Count CSIT students whose Score is 80 or more
(
    (df["Department"] == "CSIT") &
    (df["Score"] >= 80)
).sum()
```

```python
# QUESTION:
# Count students who are from Pokhara and have Age greater than 20.

# Write your answer below
# SOLUTION:
(
    (df["City"] == "Pokhara") &
    (df["Age"] > 20)
).sum()

```

### 11. Count Groups with `groupby().count()`

`groupby().count()` counts non-missing values inside each group.


```python
# Count available Score values inside each Department
df.groupby("Department")["Score"].count()
```

```python
# QUESTION:
# Count available City values inside each Department.

# Write your answer below
# SOLUTION:
df.groupby("Department")["City"].count()

```

### 12. Count Total Rows per Group with `groupby().size()`

`size()` counts every row inside each group.

This is different from `count()`, because `count()` ignores missing values.


```python
# Count total students in each Department
df.groupby("Department").size()
```

```python
# QUESTION:
# Count total rows for each City using groupby().size().

# Write your answer below
# SOLUTION:
df.groupby("City").size()

```

### Counting Patterns to Remember

`len(df)`  
→ total rows

`df["column"].count()`  
→ non-missing values

`df["column"].value_counts()`  
→ frequency of each value

`df["column"].nunique()`  
→ number of unique values

`(condition).sum()`  
→ count rows matching a condition

`df.groupby("column").size()`  
→ total rows in each group

`df.groupby("column")["value"].count()`  
→ non-missing values inside each group

`df.isna().sum()`  
→ count missing values

`pd.crosstab(col1, col2)`  
→ frequency table
