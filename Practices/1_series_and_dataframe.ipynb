#Pandas
Pandas is a Python library used for working with and analyzing data. It provides two main data structures: Series and DataFrame.

Series → 1-dimensional data
DataFrame → 2-dimensional data with rows and columns

#Detailed Comparison Between Series and DataFrame
In pandas, Series and DataFrame are two important data structures used for storing and analyzing data. A Series is a one-dimensional data structure that contains a single sequence of values along with an index. It is similar to a single column in a table. For example, a Series can be used to store the marks of students. On the other hand, a DataFrame is a two-dimensional data structure that contains rows and columns, similar to a table in Excel. It can store multiple types of information such as student name, age, marks, and address in different columns. A Series has only one dimension, while a DataFrame has two dimensions. A DataFrame can also contain multiple Series, where each column represents a Series. Therefore, Series is mainly used when working with a single column or a single set of data, whereas DataFrame is used when working with a complete dataset containing multiple rows and columns.

#Series vs DataFrame Code Comparison

```python
marks = pd.Series([75, 82, 68, 90])

print(marks)
print(marks.shape)
```

#dataframe

```python
data = {
    "Name": ["Ram", "Sita", "Hari", "Gita"],
    "Marks": [75, 82, 68, 90]
}

df = pd.DataFrame(data)

print(df)
print(df.shape)
```
