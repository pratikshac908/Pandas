````python
# 1️⃣ Import Pandas Library
# Import pandas to work with DataFrames
import pandas as pd
````

````python
# 2️⃣ Create a Sample DataFrame
# We will create a DataFrame with columns 'Name', 'Age', 'City'
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Age': [25, 30, 35, 40],
    'City': ['Kathmandu', 'Pokhara', 'Lalitpur', 'Biratnagar']
}

df = pd.DataFrame(data)  # Create DataFrame
df
````

````python
# 3️⃣ Default Index
# By default, Pandas assigns a numeric index starting from 0
print(df.index)  # Shows the default index
````

# 4️⃣ Set a Column as Index

````python
df.set_index("Name")
````

````python
df.set_index("Name",inplace=True)
````

````python
# 6️⃣ Reset Index

df.reset_index()

````

````python
# 6️⃣ Reset Index

df.reset_index(inplace=True)

````

### What does inplace=True mean?

- `inplace=True` means **modify the original DataFrame**
- No new DataFrame is created
- Changes happen **permanently**

# Pandas Index Practice Task

Practice **default index, set index, reset index, and inplace=True** using one dataset.

## Dataset
Student information with Student_ID, Name, Age, Department, Marks, and City.

````python
# 1. Import pandas
import pandas as pd
````

````python
# 2. Create the dataset
data = {
    'Student_ID': [101, 102, 103, 104, 105, 106],
    'Name': ['Ram', 'Sita', 'Hari', 'Gita', 'Asha', 'Bikash'],
    'Age': [20, 21, 19, 22, 20, 21],
    'Department': ['IT', 'BCA', 'BIM', 'IT', 'BCA', 'BIM'],
    'Marks': [85, 90, 78, 92, 88, 75],
    'City': ['Kathmandu', 'Pokhara', 'Lalitpur', 'Bhaktapur', 'Chitwan', 'Butwal']
}

df = pd.DataFrame(data)
df
````

## Task 1: Check the Default Index

````python
# Write code to display the index of df
df.index
````

## Task 2: Set Student_ID as the Index (without inplace)

**Goal:** Create a new DataFrame named df_indexed.

````python
df_indexed = df.set_index('Student_ID')
df_indexed
````

## Task 3: Test whether the original df changed

If inplace=False (default), the original DataFrame should still have the normal numeric index.

````python
# Check the original df
df
````

## Task 4: Set Student_ID as the Index with inplace=True

Now change the **original** df.

````python
df.set_index('Student_ID', inplace=True)
df
````

## Task 5: Test inplace=True

Check whether Student_ID is now the index of the original df.

````python
df.index
````

## Task 6: Reset the Index (without inplace)

Create a new DataFrame named df_reset.

````python
df_reset = df.reset_index()
df_reset
````

## Task 7: Test the original df again

Because inplace was not used in Task 6, the original df should still have Student_ID as its index.

````python
df
````

## Task 8: Reset the Index with inplace=True

Now permanently reset the index of the original df.

````python
df.reset_index(inplace=True)
df
````
